# L1 / L2 Regularization Tutorial

Compare Linear Regression, Ridge (L2), and Lasso (L1).
Always scale features before regularization.

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_squared_error
```

## Load and prepare data (replace with your dataset)

```python
# Example: df = pd.read_csv("dataset.csv")
# X = df.drop("target", axis=1)
# y = df["target"]

# For demo, create synthetic data with multicollinearity
np.random.seed(42)
n = 200
x1 = np.random.randn(n)
x2 = x1 + np.random.randn(n) * 0.1   # highly correlated
x3 = np.random.randn(n)
y = 3*x1 + 2*x2 + 0.5*x3 + np.random.randn(n)*0.5

X = pd.DataFrame({"x1": x1, "x2": x2, "x3": x3})
print(X.corr())
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.25, random_state=42
)
```

## Linear vs Ridge vs Lasso

```python
models = {
    "Linear": LinearRegression(),
    "Ridge (alpha=1)": Ridge(alpha=1.0),
    "Ridge (alpha=10)": Ridge(alpha=10.0),
    "Lasso (alpha=0.1)": Lasso(alpha=0.1),
    "Lasso (alpha=0.5)": Lasso(alpha=0.5),
}

for name, model in models.items():
    model.fit(X_train, y_train)
    r2 = model.score(X_test, y_test)
    print(f"{name:20} R²={r2:.3f}  coefs={np.round(model.coef_, 3)}")
```

## Observations
- Linear regression coefficients can be large and unstable with correlated features.
- Ridge shrinks them.
- Lasso can zero some coefficients out (feature selection).
