# SVM – Weather Classification (Linear vs RBF)

Based on codebasics exercise. Compare linear and RBF kernels with pipelines.

```python
import warnings
warnings.filterwarnings("ignore")
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, accuracy_score
```

```python
df = pd.read_csv("weather_classification_data.csv")
print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
# Adjust target column name if needed based on the actual CSV
target_col = [c for c in df.columns if "weather" in c.lower() or c == df.columns[-1]][0]
X = df.drop(target_col, axis=1)
y = df[target_col]

cat_cols = [c for c in X.columns if X[c].dtype == "object"]
num_cols = [c for c in X.columns if c not in cat_cols]

print("Categorical:", cat_cols)
print("Numerical:", num_cols)
print("Target:", target_col)
```

```python
preprocessor = ColumnTransformer([
    ("num", StandardScaler(), num_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), cat_cols)
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

## Linear Kernel SVM

```python
pipe_linear = Pipeline([
    ("prep", preprocessor),
    ("svm", SVC(kernel="linear"))
])
pipe_linear.fit(X_train, y_train)

y_pred_lin = pipe_linear.predict(X_test)
print("Linear accuracy:", accuracy_score(y_test, y_pred_lin))
print(classification_report(y_test, y_pred_lin))
```

## RBF Kernel SVM (custom hyperparameters)

```python
pipe_rbf = Pipeline([
    ("prep", preprocessor),
    ("svm", SVC(kernel="rbf", C=0.5, gamma="auto"))
])
pipe_rbf.fit(X_train, y_train)

y_pred_rbf = pipe_rbf.predict(X_test)
print("RBF accuracy:", accuracy_score(y_test, y_pred_rbf))
print(classification_report(y_test, y_pred_rbf))
```
