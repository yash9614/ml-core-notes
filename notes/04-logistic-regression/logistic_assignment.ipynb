# Logistic Regression – Multiclass Dry Bean Classification

Based on codebasics exercise. Predict bean class from geometric features.

```python
import warnings
warnings.filterwarnings("ignore")
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```python
# Load data (place dry_bean_dataset.xlsx in the same folder)
df = pd.read_excel("dry_bean_dataset.xlsx")
print(df.shape)
print(df["class"].value_counts())
df.head()
```

```python
X = df.drop("class", axis=1)
y = df["class"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

## Default Logistic Regression

```python
pipe_default = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", LogisticRegression(max_iter=1000))
])
pipe_default.fit(X_train, y_train)

y_pred = pipe_default.predict(X_test)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

## Logistic Regression with custom parameters

```python
pipe_custom = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", LogisticRegression(
        max_iter=300,
        C=0.5,
        class_weight="balanced",
        tol=0.001
    ))
])
pipe_custom.fit(X_train, y_train)

y_pred_custom = pipe_custom.predict(X_test)
print(classification_report(y_test, y_pred_custom))
print(confusion_matrix(y_test, y_pred_custom))
```
